To display or return the strings, .groups() returns a tuple of all of the gorups in the match. .findall() returns a list of tuples when used  with groups.

The findall() returns all non-overlapping matches of a pattern in a string asa  lsit of strings. As opposed to re.search(), findall() scans the entire string

```python
import re
from pathlib import Path
```

```python
input_path =Path("mbox-short.txt")
output_path = Path("ip-addresses.txt")
```

```python
octet = r"(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]?\d)"

ip_pattern = re.compile(
    rf"(?<![\d.]){octet}(?:\.{octet}){{3}}(?![\d.])"
)
```

```python
ip_addresses = []
with input_path.open("r", encoding="utf-8") as input_file:
    for line in input_file:
        for match in ip_pattern.finditer(line):
            ip_addresses.append(match.group())

with output_path.open("w", encoding="utf-8") as output_file:
    for address in ip_addresses:
        output_file.write(address + "\n")
print(f"Wrote {len(ip_addresses)} IP addresses to {output_path}")
```
